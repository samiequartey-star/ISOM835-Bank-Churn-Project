## Address Outliers and Anomalies

### Subtask:
Identify potential outliers in numerical features using the IQR method and visualize them, then apply a capping strategy for 'Age' and 'Balance'.


**Reasoning**:
The first step is to calculate the first quartile (Q1), third quartile (Q3), and the Interquartile Range (IQR) for the specified numerical features, and then use these to define upper and lower bounds for outlier detection. After that, I will identify and print the number and percentage of outliers for each feature.



```python
numerical_features = ['CreditScore', 'Age', 'Tenure', 'Balance', 'NumOfProducts', 'EstimatedSalary']

print("Outlier Analysis (IQR Method):")
for feature in numerical_features:
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[feature] < lower_bound) | (df[feature] > upper_bound)]
    num_outliers = len(outliers)
    percentage_outliers = (num_outliers / len(df)) * 100

    print(f"\n--- {feature} ---")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Number of outliers: {num_outliers}")
    print(f"Percentage of outliers: {percentage_outliers:.2f}%")
```

**Reasoning**:
The next step is to create box plots for 'Age' and 'Balance' to visually confirm the presence and extent of outliers, as specified by the subtask. This will help in understanding the distribution and impact of outliers before applying capping.



```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['Age'], palette='viridis')
plt.title('Box Plot of Age')
plt.ylabel('Age')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['Balance'], palette='viridis')
plt.title('Box Plot of Balance')
plt.ylabel('Balance')

plt.tight_layout()
plt.show()
```

The warnings do not affect the functionality but indicate a best practice for future seaborn versions. The next logical step, following visualization, is to apply the capping strategy for 'Age' and 'Balance' as specified by the subtask instructions. This involves recalculating the upper bounds for these features and then replacing values exceeding these bounds with the upper bound itself. This will be done in a single code block to complete the 'Address Outliers and Anomalies' subtask.



```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['Age'], palette='viridis', legend=False)
plt.title('Box Plot of Age')
plt.ylabel('Age')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['Balance'], palette='viridis', legend=False)
plt.title('Box Plot of Balance')
plt.ylabel('Balance')

plt.tight_layout()
plt.show()

# Apply capping for 'Age' and 'Balance' based on IQR upper bound
print("\nApplying capping to 'Age' and 'Balance'...")

# Calculate bounds for Age
Q1_age = df['Age'].quantile(0.25)
Q3_age = df['Age'].quantile(0.75)
IQR_age = Q3_age - Q1_age
upper_bound_age = Q3_age + 1.5 * IQR_age

# Cap 'Age' values
df['Age'] = df['Age'].clip(upper=upper_bound_age)
print(f"Age values capped above {upper_bound_age:.2f}")

# Calculate bounds for Balance
Q1_balance = df['Balance'].quantile(0.25)
Q3_balance = df['Balance'].quantile(0.75)
IQR_balance = Q3_balance - Q1_balance
upper_bound_balance = Q3_balance + 1.5 * IQR_balance

# Cap 'Balance' values
df['Balance'] = df['Balance'].clip(upper=upper_bound_balance)
print(f"Balance values capped above {upper_bound_balance:.2f}")

print("Capping complete.")
```
